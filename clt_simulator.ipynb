```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
# Skewed Exponential Population (lambda = 1.0)
population = np.random.exponential(scale=1.0, size=100000)

# Sampling Means for n=5 and n=50 (1000 trials each)
means_n5 = [np.mean(np.random.choice(population, 5)) for _ in range(1000)]
means_n50 = [np.mean(np.random.choice(population, 50)) for _ in range(1000)]

fig, axs = plt.subplots(1, 3, figsize=(15, 4.5))

axs[0].hist(population, bins=50, color='crimson', alpha=0.7, density=True)
axs[0].set_title("1. Skewed Population (Exponential)")
axs[0].set_xlim(0, 6)

axs[1].hist(means_n5, bins=30, color='darkorange', alpha=0.8, density=True)
axs[1].set_title(f"2. Sample Means (n=5, SE={np.std(means_n5):.3f})")

axs[2].hist(means_n50, bins=30, color='royalblue', alpha=0.8, density=True)
axs[2].axvline(np.mean(population), color='black', linestyle='--', label='True Mean (1.0)')
axs[2].set_title(f"3. Sample Means (n=50, SE={np.std(means_n50):.3f})")
axs[2].legend()

plt.tight_layout()
plt.savefig("clt_simulation.png", dpi=300)
plt.show()
```
